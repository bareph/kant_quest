```python
# 부모 클래스 Base Class, Super Class, Parent Class
class Animal:
    # 공통 기능
    def breath(self):
        print('숨을 쉽니다.')
    
    def sound(self):
        print('소리를 냅니다.')
    
# 상속(Inheritance)
class Dog(Animal):
    # 상속 받은 클래스에서만 있는 기능. 고유 기능
    def bark(self):
        print('멍멍')

# 인스턴스 생성    
dog1 = Dog()

dog1.breath()   # 부모 클래스로부터 입력 받은 메서드를 그대로 사용할 수 있다.
dog1.bark()     # 자식 클래스의 고유 기능

# 메서드 오버라이딩(Method Overriding)
# 부모 클래스에서 정의된 메서드를 수정 또는 추가하는 기능(재정의)

class Cat(Animal):
    def sound(self):
        print('애옹') # 재정의
    
nabi = Cat()
nabi.sound()
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, major):
        super().__init__(name, age)  # super()를 통해서 부모 클래스의 생성자를 접근해서 name 속성을 설정한다.
        self.major = major

std1 = Student('최진웅', 32, '스포츠댄스')

print(std1.name)    # 부모 클래스로부터 물려받은 속성
print(std1.major)   # 자식 고유의 속성
```

```python
class Wallet:
    def __init__(self):
        self.__money = 1000 # 숨겨진 데이터
    
    def get_money(self):    # Getter : 값을 안전하게 전달하는 용도
        return self.__money

    def set_money(self, amount) : # Setter : 검색 로직을 추가
        if amount >= 0:
            self.__money = amount
        else:
            print('잘못된 값입니다.')

wallet = Wallet()
wallet.set_money(2000)
wallet.set_money(3000)
print(wallet.get_money())
```
